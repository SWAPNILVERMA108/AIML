```python
oscar_movies=[
    "The Dark Knight",
    "The Hurt Locker",
    "The Kings Speech",
    "The Artist",
    "Argo"
]
year =[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]

plt.plot(year,oscar_revenue)
plt.title("oscar movies revenue in each year")
plt.xlabel("year")
plt.ylabel("Reveneue ( M $)")
plt.plot()
```

```python
with plt.xkcd():
    non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo","Lincoln"]
    non_oscar_revenue = [378, 2788, 829, 185, 275]
    plt.plot(year,oscar_revenue,"o--b", label="oscar_movies")
    plt.plot(year,non_oscar_revenue ,">--r",label="non-oscar movies")
    plt.title("oscar movie revenue vs non-oscar movie revenue")
    plt.xlabel("year")
    plt.ylabel("revenue in M $")
    plt.plot()
    plt.legend()
```

```python
oscar_movies=[
    "The Dark Knight",
    "The Hurt Locker",
    "The Kings Speech",
    "The Artist",
    "Argo"
]
year =[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]

plt.bar(year,oscar_revenue,color="yellow")
plt.title("oscar movies revenue in each year")
plt.xlabel("year")
plt.ylabel("Reveneue ( M $)")
plt.plot()

for i in range(len(year)):
    plt.text(year[i],oscar_revenue[i]+15,str(oscar_revenue[i]),ha="center")
    
    
# for increase the length of graph 

plt.ylim(0,max(oscar_revenue)+100)
```
