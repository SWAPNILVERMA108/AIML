```python
oscar_movies=[
    "The Dark Knight",
    "The Hurt Locker",
    "The Kings Speech",
    "The Artist",
    "Argo"
]
year =[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]

plt.bar(year,oscar_revenue,color="yellow")
plt.title("oscar movies revenue in each year")
plt.xlabel("year")
plt.ylabel("Reveneue ( M $)")
plt.plot()

for i in range(len(year)):
    plt.text(year[i],oscar_revenue[i]+15,str(oscar_revenue[i]),ha="center")
    
    
# for increase the length of graph 

plt.ylim(0,max(oscar_revenue)+100)
```

```python
import numpy as np
oscar_movies=[
    "The Dark Knight",
    "The Hurt Locker",
    "The Kings Speech",
    "The Artist",
    "Argo"
]
year =[2008,2009,2010,2011,2012]
oscar_revenue=[1005,170,427,133,232]

non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo","Lincoln"]
non_oscar_revenue =[378,2788,829,185,275]


plt.title("oscar movie vs non oscar movie")
plt.xlabel("year")
plt.ylabel("Revenue (M $)")

x = np.arange(len(year))

width =0.4

plt.bar(x-width/2,oscar_revenue,width,label="oscar_movies")
plt.bar(x+width/2,non_oscar_revenue,width,label="non_oscar_movies")
plt.legend()
```

```python
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
"Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors= ["green" if x <135 else "red" for x in blood_pressure]
plt.scatter(age,blood_pressure,color=colors,s=blood_pressure,marker=">",alpha=0.7,cmap=)
plt.title("Age Vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
```

Seaborn tutorial

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
fmri=sns.load_dataset("fmri")
sns.lineplot(x="timepoint",y="signal",hue="region",data=fmri)
plt.show()
```

```python
tips = sns.load_dataset("tips")
sns.scatterplot(x="total_bill",y="tip",hue="day",data=tips)
plt.show()
```

```python
tips = sns.load_dataset("tips")
sns.barplot(x="day",y="total_bill",data=tips)
plt.show()
tips.head()
```

```python
tips = sns.load_dataset("tips")
sns.boxplot(x="day",y="total_bill",data=tips)
plt.show()
```

```python
sns.histplot(tips["total_bill"],kde=True)
plt.show()
```

```python
corr  =tips.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
```

```python
iris = sns.load_dataset("iris")
sns.pairplot(iris,hue="species")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

x = ['sun', 'mon', 'fri', 'sat', 'tue', 'wed', 'thu']
y = [5, 6.7, 4, 6, 2, 4.9, 1.8]

ax = sns.stripplot(x=x, y=y)
ax.set(xlabel='Days', ylabel='Amount Spent')
plt.title('Daily Spending (Custom Data)')

plt.grid()
plt.show()
```

```python
sns.set(style="whitegrid")
iris = sns.load_dataset("iris")
iris.head()
sns.swarmplot(x="species",y="sepal_length",data=iris)
plt.title("swarm plot of speal length by species")
plt.show()
```
